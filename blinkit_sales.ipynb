```python
blinkit_orders['promised_delivery_time']=pd.to_datetime(blinkit_orders['promised_delivery_time'])
blinkit_orders['actual_delivery_time']=pd.to_datetime(blinkit_orders['actual_delivery_time'])
blinkit_orders['order_id']=blinkit_orders['order_id'].astype(str)
```

```python
#calculate delay minutes
blinkit_orders['delay_minutes']=( blinkit_orders['actual_delivery_time']-blinkit_orders['promised_delivery_time']).dt.total_seconds()/60
```

```python
def check_status(x):
    if x<=5:
        return 'On Time'
    elif x<=15:
        return 'Slightly Delayed'
    else:
        return 'Significantly Delayed'
    

blinkit_orders['new_delivery_status']=(blinkit_orders['delay_minutes'].apply(check_status))
```
